### 02 · 变量

变量是内存中用来存储数据、并可通过名称访问的"容器"。它的值在程序运行期间可以被改变。



#### 变量声明语法

数据类型 变量名 = 初始值；

> **要点**：声明变量必须指定类型；变量须先赋值后才能使用，否则编译报错。



````
int age = 18;           // 整数
double price = 9.99;    // 浮点数
string name = "小明";   // 字符串
bool isOk = true;       // 布尔

// 也可以先声明，再赋值
int score;
score = 95;
````

#### 整数类型

### 03 · 字符串




#### 什么是字符串？

字符串（string）是由零个或多个字符组成的文本序列，用双引号包裹，是编程中最常用的数据类型之一。



#### 字符串声明与长度

声明字符串、获取长度、按索引取字符。

> **要点**：字符串是不可变的；Length 获取长度，[索引] 访问单个字符。



````
string name = "小明";
string empty = "";              // 空字符串
string path = @"C:\temp\file";  // 逐字字符串 @

// 字符串长度
int len = name.Length;          // 2

// 按索引访问单个字符
char first = name[0];          // '小'
````

#### 字符串拼接

用 + 或 Concat 把多个字符串连在一起。

> **要点**：用 + 拼接最直观；数字与字符串相加会自动转换为文本。



````
string a = "Hello";
string b = "World";

// 用 + 拼接
string c = a + " " + b;          // "Hello World"

// 用 string.Concat 拼接
string d = string.Concat(a, " ", b);

// 数字与字符串拼接会自动转文本
int n = 42;
string s = "答案是 " + n;     // "答案是 42"
````

#### 字符串插值 $

用 $ 与花括号把变量直接嵌入文本。

> **要点**：$"..." 中的 { } 可嵌入变量或表达式，比 + 拼接更清晰易读。



````
string name = "小明";
int age = 18;

// 字符串插值：$ 前缀 + {变量}
string s = $"我叫{name}，今年{age}岁。";

// 花括号内还能嵌入表达式
string t = $"明年我{age + 1}岁。";

Console.WriteLine(s);
````

#### 转义字符

用反斜杠 \ 表示特殊字符。



#### 字符串常用方法

这些方法帮你轻松完成文本处理。



#### 查找与比较

定位子串、截取片段、比较相等。

> **要点**：字符串比较用 == 或 Equals，默认区分大小写。



````
string s = "Hello";

bool has = s.Contains("ell");     // true
int idx = s.IndexOf("l");         // 2
string sub = s.Substring(1, 3);  // "ell"
string rep = s.Replace("l", "L"); // "HeLLo"

// 相等比较
bool eq = s == "Hello";          // true
````

### 04 · 条件语句




#### 条件语句概述

条件语句让程序能够"做判断"，根据不同的条件选择不同的执行路径，是程序逻辑的核心。



#### if 语句

条件成立时，才执行代码块。

> **要点**：if 后面的条件为 true 时，才执行花括号内的代码。



````
int score = 85;

if (score >= 60)
{
    Console.WriteLine("及格了！");
}
````

#### if-else 二选一

条件成立走 if，否则走 else。

> **要点**：二选一的判断：if 条件成立执行 if 块，否则执行 else 块。



````
int score = 85;

if (score >= 60)
{
    Console.WriteLine("及格");
}
else
{
    Console.WriteLine("不及格");
}
````

#### else if 多分支

依次判断多个条件，命中即停。

> **要点**：多分支依次判断，命中一个后其余不再执行；单条语句可省略花括号。



````
int score = 85;

if (score >= 90)
    Console.WriteLine("优秀");
else if (score >= 80)
    Console.WriteLine("良好");
else if (score >= 60)
    Console.WriteLine("及格");
else
    Console.WriteLine("不及格");
````

#### 嵌套 if

if 里面再套 if，处理复合条件。

> **要点**：if 可以嵌套，注意花括号配对与缩进，保持代码清晰。



````
bool isLogin = true;
int score = 85;

if (isLogin)
{
    if (score >= 90)
        Console.WriteLine("优秀");
    else
        Console.WriteLine("继续努力");
}
else
{
    Console.WriteLine("请先登录");
}
````

#### 三元运算符 ?:

一行写出简单的 if-else 判断。

> **要点**：条件 ? 值1 : 值2，条件为真返回"值1"，否则返回"值2"。



````
int score = 85;

string result = score >= 60 ? "及格" : "不及格";

Console.WriteLine(result);

// 等价于：
// if (score >= 60) result = "及格";
// else             result = "不及格";
````

#### switch 语句

对同一变量做多值匹配。

> **要点**：switch 按值匹配 case，default 处理未匹配情况，break 结束分支。



````
int day = 3;

switch (day)
{
    case 1:
        Console.WriteLine("星期一");
        break;
    case 2:
        Console.WriteLine("星期二");
        break;
    default:
        Console.WriteLine("其他");
        break;
}
````

#### switch 表达式

C# 8+ 更简洁的 switch 写法。

> **要点**：switch 表达式用 => 直接返回结果，_ 是默认匹配，无需 break。



````
int day = 3;

string name = day switch
{
    1 => "星期一",
    2 => "星期二",
    3 => "星期三",
    _ => "其他"
};

Console.WriteLine(name);
````

### 05 · 循环




#### 循环概述

循环让程序重复执行一段代码，避免重复书写，是处理批量数据与迭代逻辑的利器。



#### for 循环

适合已知次数的计数循环。

> **要点**：三部分依次为：初始化、循环条件、迭代更新；条件为 false 时结束。



````
// for (初始化; 条件; 迭代)
for (int i = 0; i < 5; i++)
{
    Console.WriteLine(i);
}
// 输出：0 1 2 3 4
````

#### while 循环

先判断条件，成立才执行。

> **要点**：先判断后执行，条件为 true 才进入循环体；别忘了在循环体内更新条件变量。



````
int i = 0;

while (i < 5)
{
    Console.WriteLine(i);
    i++;
}
````

#### do-while 循环

先执行一次，再判断条件。

> **要点**：先执行一次循环体再判断，保证至少执行一次。



````
int i = 0;

do
{
    Console.WriteLine(i);
    i++;
} while (i < 5);
````

#### foreach 循环

遍历数组或集合中的每个元素。

> **要点**：foreach 自动取出集合中每个元素，简单安全，适合只读遍历。



````
int[] nums = { 10, 20, 30 };

foreach (int n in nums)
{
    Console.WriteLine(n);
}
// 输出：10 20 30
````

#### break 与 continue

控制循环的跳出与跳过。

> **要点**：continue 跳过本次剩余代码继续下一轮；break 直接终止整个循环。



````
for (int i = 0; i < 10; i++)
{
    if (i == 3) continue;  // 跳过本次
    if (i == 7) break;     // 终止循环
    Console.WriteLine(i);
}
// 输出：0 1 2 4 5 6
````

#### 循环嵌套

循环套循环，打印九九乘法表。

> **要点**：内层循环完整执行一轮，外层循环才前进一步；注意控制层数避免逻辑混乱。



````
for (int i = 1; i <= 9; i++)
{
    for (int j = 1; j <= i; j++)
    {
        Console.Write($"{j}×{i}={i * j}\t");
    }
    Console.WriteLine();
}
````

### 06 · 函数




#### 什么是函数？

函数（方法）是一段可复用的代码块，把一组操作封装起来，通过名称即可重复调用。



#### 函数定义语法

返回类型 + 方法名 + 参数列表 + 方法体。

> **要点**：方法定义在类中；调用时使用 方法名()。



````
// 返回类型 方法名(参数列表) { 方法体 }
static void SayHello()
{
    Console.WriteLine("你好！");
}

// 调用方法
SayHello();
````

#### void 无返回值

只执行操作、不返回结果的方法。

> **要点**：void 表示方法不返回任何值，仅执行其中的语句。



````
static void ShowMenu()
{
    Console.WriteLine("1. 开始游戏");
    Console.WriteLine("2. 退出");
}

ShowMenu();
````

#### 带参数的方法

通过参数向方法传递数据。

> **要点**：形参写在括号内；调用时传入实参，按位置一一对应。



````
static void Greet(string name)
{
    Console.WriteLine($"你好，{name}！");
}

Greet("小明");
Greet("小红");
````

#### 返回值 return

用 return 把计算结果交还给调用方。

> **要点**：返回类型写在方法名前；return 返回结果并结束方法执行。



````
static int Add(int a, int b)
{
    return a + b;
}

int sum = Add(3, 5);   // 8
Console.WriteLine(sum);
````

#### 多参数与命名参数

传多个参数，或用名称指定参数。

> **要点**：命名参数用 名称: 值 传入，可忽略顺序，提高可读性。



````
static double Calc(double price, int count, double discount)
{
    return price * count * discount;
}

// 位置参数（按顺序传入）
double r1 = Calc(10.0, 3, 0.8);

// 命名参数（可打乱顺序）
double r2 = Calc(count: 3, discount: 0.8, price: 10.0);
````

#### 可选参数与默认值

给参数设默认值，调用时可省略。

> **要点**：可选参数需给默认值；默认参数必须放在必选参数之后。



````
static void Greet(string name, string title = "同学")
{
    Console.WriteLine($"你好，{title}{name}！");
}

Greet("小明");               // 你好，同学小明！
Greet("小红", "老师");       // 你好，老师小红！
````

#### 方法重载

同名方法，参数列表不同。

> **要点**：重载 = 方法名相同、参数列表（个数或类型）不同；仅返回类型不同不构成重载。



````
static int Add(int a, int b) { return a + b; }

static double Add(double a, double b) { return a + b; }

static int Add(int a, int b, int c) { return a + b + c; }

Console.WriteLine(Add(1, 2));      // int 版本
Console.WriteLine(Add(1.5, 2.5));  // double 版本
Console.WriteLine(Add(1, 2, 3));   // 三参数版本
````

#### 方法与作用域

理解方法内部变量的可见范围与生命周期。



#### 递归调用

方法自己调用自己，解决可分解问题。

> **要点**：递归必须包含终止条件，否则会导致无限调用、栈溢出。



````
// 计算 n 的阶乘 n!
static int Factorial(int n)
{
    if (n <= 1) return 1;   // 终止条件
    return n * Factorial(n - 1);
}

Console.WriteLine(Factorial(5));  // 120
````

#### 值传递 vs 引用传递

默认按值传递，ref 可按引用传递。



#### ref 与 out 参数

让方法能修改外部变量或输出结果。

> **要点**：ref 双向传递；out 用于仅输出，方法内必须先赋值。



````
// ref：双向传递，实参须先初始化
static void Swap(ref int a, ref int b)
{
    int tmp = a; a = b; b = tmp;
}

int x = 1, y = 2;
Swap(ref x, ref y);   // x=2, y=1
````

### 07 · 类




#### 面向对象思想

面向对象（OOP）把数据与操作数据的方法打包成"对象"，让程序更贴近现实世界、更易维护。



#### 类与对象

类是对象的"蓝图/模板"，对象是类的具体实例。用 new 关键字根据类创建出对象。



#### 类的定义

用 class 定义字段（数据）与方法（行为）。

> **要点**：class 关键字定义类；类内可包含字段（数据）和方法（行为）。



````
class Person
{
    // 字段（数据）
    public string name;
    public int age;

    // 方法（行为）
    public void SayHello()
    {
        Console.WriteLine($"我是{name}，今年{age}岁。");
    }
}
````

#### 字段 Field

字段保存对象的状态数据。

> **要点**：字段是类中的变量，每个对象都拥有自己的一份字段数据。



````
class Person
{
    public string name;   // 字段：保存姓名
    public int age;       // 字段：保存年龄

    public void Introduce()
    {
        Console.WriteLine($"{name}, {age} 岁");
    }
}
````

#### 属性 Property

用 get / set 封装字段的读写。

> **要点**：属性通过 get/set 访问器封装字段，value 是传入的新值。



````
class Person
{
    private int age;         // 私有字段

    public int Age            // 属性
    {
        get { return age; }
        set { age = value; }
    }
}
````

#### 自动属性

简化写法，编译器自动生成后备字段。

> **要点**：自动属性省略方法体，编译器自动生成后备字段，代码更简洁。



````
class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
}

var p = new Person();
p.Name = "小明";
p.Age = 18;
Console.WriteLine(p.Name);
````

#### 构造函数

与类同名，new 时自动执行的初始化方法。

> **要点**：构造函数在 new 时自动执行，用于初始化对象的初始状态。



````
class Person
{
    public string Name;
    public int Age;

    // 构造函数：与类同名，无返回类型
    public Person(string name, int age)
    {
        Name = name;
        Age = age;
    }
}

var p = new Person("小明", 18);
````

#### 类中的方法

方法描述对象的行为，通过对象调用。

> **要点**：类中的方法描述对象的行为，通过 对象.方法() 调用。



````
class Calculator
{
    public int Add(int a, int b)
    {
        return a + b;
    }
}

var c = new Calculator();
int r = c.Add(3, 5);   // 8
````

#### 创建对象并访问成员

new 实例化，用 . 访问成员。

> **要点**：用 new 实例化；通过 对象.成员 访问字段、属性与方法。



````
class Person
{
    public string Name { get; set; }

    public void SayHello()
    {
        Console.WriteLine($"你好，我是{Name}。");
    }
}

var p = new Person();
p.Name = "小明";      // 访问属性
p.SayHello();      // 调用方法
````

#### 访问修饰符

控制类成员可被哪些代码访问。



#### 封装

隐藏内部细节，只暴露必要的访问接口。



#### 静态成员 static

属于类本身，无需实例化即可使用。

> **要点**：static 成员属于类本身，通过 类名.成员 访问，不需要实例化。



````
class MathUtil
{
    public static double PI = 3.14159;

    public static double Area(double r)
    {
        return PI * r * r;
    }
}

// 静态成员用 类名.成员 访问，无需 new
Console.WriteLine(MathUtil.Area(2));
````

### 08 · 文本读写




#### 文件读写概述

文本读写让程序能持久化数据——把运行结果保存到文件，或从文件读取数据。



#### File 类入门

File 的静态方法，一步完成读写。

> **要点**：需要 using System.IO；File 的静态方法简单易用，适合小文件。



````
using System.IO;

string path = "data.txt";

File.WriteAllText(path, "你好，文件！");  // 写
string content = File.ReadAllText(path);  // 读

Console.WriteLine(content);
````

#### 读取文件

读全文，或按行读为数组。

> **要点**：ReadAllText 读全文为字符串；ReadAllLines 按行读为数组。



````
using System.IO;

string path = "data.txt";

// 一次性读取全部文本
string all = File.ReadAllText(path);

// 按行读取为字符串数组
string[] lines = File.ReadAllLines(path);

foreach (string line in lines)
{
    Console.WriteLine(line);
}
````

#### 写入与追加

覆盖写入 vs 末尾追加。

> **要点**：WriteAllText 覆盖原文件；AppendAllText 在末尾追加不覆盖。



````
using System.IO;

string path = "log.txt";

// 覆盖写入（清空原内容）
File.WriteAllText(path, "第一行\n");

// 追加写入（保留原内容）
File.AppendAllText(path, "第二行\n");

Console.WriteLine(File.ReadAllText(path));
````

#### StreamReader 逐行读

适合大文件的逐行读取。

> **要点**：ReadLine 逐行读取，读到末尾返回 null；配合 using 自动释放。



````
using System.IO;

using (StreamReader reader = new StreamReader("data.txt"))
{
    string line;
    while ((line = reader.ReadLine()) != null)
    {
        Console.WriteLine(line);
    }
}
````

#### StreamWriter 逐行写

逐行写入文本文件。

> **要点**：WriteLine 逐行写入；using 块结束自动关闭并刷新缓冲区。



````
using System.IO;

using (StreamWriter writer = new StreamWriter("out.txt"))
{
    writer.WriteLine("第一行");
    writer.WriteLine("第二行");
}
// 离开 using 块，文件自动关闭
````

#### 异常处理 try-catch

捕获文件读写错误，防止程序崩溃。

> **要点**：先捕获具体异常，再捕获通用 Exception；异常被捕获后程序继续运行。



````
using System.IO;

try
{
    string content = File.ReadAllText("不存在.txt");
    Console.WriteLine(content);
}
catch (FileNotFoundException)
{
    Console.WriteLine("文件不存在！");
}
catch (Exception ex)
{
    Console.WriteLine("出错：" + ex.Message);
}
````

#### using 语句

确保文件流等资源被及时释放。

> **要点**：using 块结束自动调用 Dispose 释放资源，避免文件句柄泄漏。



````
using System.IO;

// using 语句确保资源自动释放
using (var reader = new StreamReader("data.txt"))
{
    Console.WriteLine(reader.ReadToEnd());
}
// 离开 using 块，reader 自动 Dispose 关闭
````

#### 综合实战：成绩统计

综合运用文件读写、循环、函数统计平均分。

> **要点**：先用 WriteAllLines 写出，再用 ReadAllLines 读回并统计，完成完整闭环。



````
using System.IO;

string path = "scores.txt";
string[] data = { "小明 90", "小红 85", "小刚 72" };
File.WriteAllLines(path, data);      // 写文件

int sum = 0, n = 0;
foreach (string line in File.ReadAllLines(path))
{
    sum += int.Parse(line.Split(' ')[1]);
    n++;
}
Console.WriteLine($"平均分：{sum / (double)n}");
````
